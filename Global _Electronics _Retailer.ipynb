```python

# Convert Sales date columns to datetime

sales["Order Date"] = pd.to_datetime(
    sales["Order Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

sales["Delivery Date"] = pd.to_datetime(
    sales["Delivery Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Verify the conversion
print(sales[["Order Date", "Delivery Date"]].dtypes)

print("\nDate range:")
print("Earliest order:", sales["Order Date"].min())
print("Latest order:", sales["Order Date"].max())

print("\nMissing dates after conversion:")
print(sales[["Order Date", "Delivery Date"]].isnull().sum())
```

```python

# Inspect and clean product prices and costs

print("Before cleaning:")
print(products[["Unit Cost USD", "Unit Price USD"]].head())

for col in ["Unit Cost USD", "Unit Price USD"]:
    products[col] = (
        products[col]
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .astype(float)
    )

print("\nAfter cleaning:")
print(products[["Unit Cost USD", "Unit Price USD"]].head())

print("\nUpdated data types:")
print(products[["Unit Cost USD", "Unit Price USD"]].dtypes)
```

```python


# Convert remaining date columns

customers["Birthday"] = pd.to_datetime(
    customers["Birthday"],
    format="%m/%d/%Y",
    errors="raise"
)

stores["Open Date"] = pd.to_datetime(
    stores["Open Date"],
    format="%m/%d/%Y",
    errors="raise"
)

exchange_rate["Date"] = pd.to_datetime(
    exchange_rate["Date"],
    format="%m/%d/%Y",
    errors="raise"
)

# Verify the conversion
print("Customers Birthday:", customers["Birthday"].dtype)
print("Stores Open Date:", stores["Open Date"].dtype)
print("Exchange Rates Date:", exchange_rate["Date"].dtype)

print("\nDate samples:")
print("Birthday:", customers["Birthday"].head(3).tolist())
print("Open Date:", stores["Open Date"].head(3).tolist())
print("Exchange Rate Date:", exchange_rate["Date"].head(3).tolist())
```

```python
# Calculate Revenue, Cost, Profit, and Profit Margin

# Merge product prices and costs into Sales

sales = sales.merge(
    products[["ProductKey", "Unit Cost USD", "Unit Price USD"]],
    on="ProductKey",
    how="left",
    validate="many_to_one"
)

# Calculate financial metrics
sales["Revenue USD"] = sales["Quantity"] * sales["Unit Price USD"]
sales["Cost USD"] = sales["Quantity"] * sales["Unit Cost USD"]
sales["Profit USD"] = sales["Revenue USD"] - sales["Cost USD"]

sales["Profit Margin %"] = (
    sales["Profit USD"] / sales["Revenue USD"] * 100
)

# Verify the results
print("Sales shape after merge:", sales.shape)
print("\nFinancial calculations:")
print(
    sales[
        [
            "Quantity",
            "Unit Price USD",
            "Unit Cost USD",
            "Revenue USD",
            "Cost USD",
            "Profit USD",
            "Profit Margin %"
        ]
    ].head(10)
)

print("\nMissing values in financial columns:")
print(
    sales[
        ["Unit Price USD", "Unit Cost USD", "Revenue USD", "Cost USD", "Profit USD"]
    ].isna().sum()
)
```

```python
# Validate product matching and financial calculations

print("Sales rows:", len(sales))
print("Missing ProductKey:", sales["ProductKey"].isna().sum())

print(
    "Missing product prices:",
    sales["Unit Price USD"].isna().sum()
)

print(
    "Missing product costs:",
    sales["Unit Cost USD"].isna().sum()
)

print(
    "Negative profit rows:",
    (sales["Profit USD"] < 0).sum()
)

print(
    "Zero or negative revenue rows:",
    (sales["Revenue USD"] <= 0).sum()
)

print("\nFinancial summary:")
print(
    sales[
        ["Revenue USD", "Cost USD", "Profit USD", "Profit Margin %"]
    ].describe().round(2)
)
```

```python
# Calculate delivery days
# Delivery Days = Delivery Date − Order Date
# Calculate delivery days
sales["Delivery Days"] = (
    sales["Delivery Date"] - sales["Order Date"]
).dt.days

# Validate delivery days
print("Missing delivery dates:", sales["Delivery Date"].isna().sum())
print("Missing delivery days:", sales["Delivery Days"].isna().sum())

print(
    "Negative delivery days:",
    (sales["Delivery Days"] < 0).sum()
)

print("\nDelivery days summary:")
print(sales["Delivery Days"].describe())

print("\nSample records:")
display(
    sales[
        ["Order Date", "Delivery Date", "Delivery Days"]
    ].head(10)
)
```

```python
# customer and store data
# Check customer birthday data
print("CUSTOMER DATA")
print("Total customers:", len(customers))
print("Missing birthdays:", customers["Birthday"].isna().sum())
print("Earliest birthday:", customers["Birthday"].min())
print("Latest birthday:", customers["Birthday"].max())

# Check store opening dates
print("\nSTORE DATA")
print("Total stores:", len(stores))
print("Missing opening dates:", stores["Open Date"].isna().sum())
print("Earliest opening date:", stores["Open Date"].min())
print("Latest opening date:", stores["Open Date"].max())

# Check for future dates relative to the sales data
print("\nDATE VALIDATION")
print(
    "Birthdays after latest order date:",
    (customers["Birthday"] > sales["Order Date"].max()).sum()
)

print(
    "Store opening dates after latest order date:",
    (stores["Open Date"] > sales["Order Date"].max()).sum()
)
```

```python
# What age groups are buying from the electronics retailer?
# Add customer birthdays to Sales
sales = sales.merge(
    customers[["CustomerKey", "Birthday"]],
    on="CustomerKey",
    how="left",
    validate="many_to_one"
)

# Calculate age at the time of each order
sales["Customer Age"] = (
    sales["Order Date"].dt.year
    - sales["Birthday"].dt.year
    - (
        (sales["Order Date"].dt.month < sales["Birthday"].dt.month)
        | (
            (sales["Order Date"].dt.month == sales["Birthday"].dt.month)
            & (sales["Order Date"].dt.day < sales["Birthday"].dt.day)
        )
    ).astype(int)
)

# Validate the results
print("Sales rows:", len(sales))
print("Missing birthdays:", sales["Birthday"].isna().sum())
print("Missing customer ages:", sales["Customer Age"].isna().sum())

print("\nCustomer age summary:")
print(sales["Customer Age"].describe())

print("\nSample results:")
display(
    sales[
        ["CustomerKey", "Order Date", "Birthday", "Customer Age"]
    ].head(10)
)
```

```python
# Create age groups
age_bins = [0, 17, 25, 35, 45, 55, 65, float("inf")]

age_labels = [
    "Under 18",
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "66+"
]

sales["Customer Age Group"] = pd.cut(
    sales["Customer Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

# Review the distribution
age_group_summary = (
    sales["Customer Age Group"]
    .value_counts(sort=False)
    .rename_axis("Customer Age Group")
    .reset_index(name="Sales Lines")
)

age_group_summary["Percentage"] = (
    age_group_summary["Sales Lines"]
    / len(sales) * 100
).round(2)

print(age_group_summary.to_string(index=False))
```

```python
cleaning_log.loc[len(cleaning_log)] = [
    "Sales",
    "Customer age and age groups",
    "Merged customer birthdays, calculated age at order date, and created seven age groups. All 62,884 sales lines have valid ages and age groups."
]

display(cleaning_log.tail(3))
```

```python
# Calculate store age at the time of each sale
# Add store information to Sales
sales = sales.merge(
    stores[["StoreKey", "Open Date", "Square Meters", "Country"]],
    on="StoreKey",
    how="left",
    validate="many_to_one",
    suffixes=("", "_Store")
)

# Calculate store age in completed years at the time of each order
sales["Store Age"] = (
    sales["Order Date"].dt.year
    - sales["Open Date"].dt.year
    - (
        (sales["Order Date"].dt.month < sales["Open Date"].dt.month)
        | (
            (sales["Order Date"].dt.month == sales["Open Date"].dt.month)
            & (sales["Order Date"].dt.day < sales["Open Date"].dt.day)
        )
    ).astype(int)
)

# Validate the merge and calculation
print("Sales rows:", len(sales))
print("Missing store opening dates:", sales["Open Date"].isna().sum())
print("Missing store ages:", sales["Store Age"].isna().sum())
print("Negative store ages:", (sales["Store Age"] < 0).sum())

print("\nStore age summary:")
print(sales["Store Age"].describe())

print("\nSample records:")
display(
    sales[
        ["StoreKey", "Order Date", "Open Date", "Store Age"]
    ].head(10)
)
```

```python
cleaning_log.loc[len(cleaning_log)] = [
    "Sales",
    "Store age",
    "Merged store opening dates and square meters; calculated store age at order date. All 62,884 rows retained, with no missing or negative store ages."
]

display(cleaning_log.tail(3))
```

```python
# Remove exact duplicate entries from the cleaning log
cleaning_log = cleaning_log.drop_duplicates().reset_index(drop=True)

display(cleaning_log.tail(5))
print("Total cleaning log entries:", len(cleaning_log))
```

```python
# Group stores by age at the time of each sale
store_age_bins = [-1, 5, 10, 15, float("inf")]

store_age_labels = [
    "0-5 years",
    "6-10 years",
    "11-15 years",
    "16+ years"
]

sales["Store Age Group"] = pd.cut(
    sales["Store Age"],
    bins=store_age_bins,
    labels=store_age_labels
)

# Summarize the distribution
store_age_summary = (
    sales["Store Age Group"]
    .value_counts(sort=False)
    .rename_axis("Store Age Group")
    .reset_index(name="Sales Lines")
)

store_age_summary["Percentage"] = (
    store_age_summary["Sales Lines"] / len(sales) * 100
).round(2)

print(store_age_summary.to_string(index=False))
```

```python
cleaning_log.loc[len(cleaning_log)] = [
    "Sales",
    "Store age groups",
    "Created four store-age groups using store age at order date; 62,884 sales lines classified."
]

display(cleaning_log.tail(3))
```
